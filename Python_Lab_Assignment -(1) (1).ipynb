## Question 8

**Find area of a triangle when lengths of all three sides are given (Heron's formula).**

🎯 Aim

Write a Python program to calculate the area of a triangle when the lengths of all three sides are given using Heron's Formula.

🌍 Real-Life Example

Imagine a civil engineer needs to calculate the area of a triangular plot of land. The three sides of the plot are 3 m, 4 m, and 5 m.

````python
import math

side1 = 3
side2 = 4
side3 = 5

s = (side1 + side2 + side3) / 2

land_area = math.sqrt(s * (s - side1) * (s - side2) * (s - side3))

print("Area of the Land:", land_area, "square meters")
````

💡 Explanation

🏗️ A civil engineer measures a triangular piece of land with sides:

📏 Side 1 = 3 meters
📏 Side 2 = 4 meters
📏 Side 3 = 5 meters

Instead of measuring the height of the triangle, the engineer uses Heron's Formula to calculate the area.

🎯 Aim

Create a Python program to implement Operator Overloading by adding two Point objects.

````python
class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"


p1 = Point(8, 14)
p2 = Point(6, 11)
p3 = p1 + p2

print("P1:", p1)
print("P2:", p2)
print("P3 = P1 + P2:", p3)
````
